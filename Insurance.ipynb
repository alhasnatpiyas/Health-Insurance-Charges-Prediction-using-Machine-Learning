**4. Feature Extraction**

```python
from scipy.stats import pearsonr
selected_features = [
    'age','bmi','children','is_female','is_smoker','region_northwest','region_southeast','region_southwest',
    'bmi_category_Normal','bmi_category_Overweight','bmi_category_Obese'
]
correlations = {
    feature: pearsonr(df_cleaned[feature], df_cleaned['charges'])[0]
    for feature in selected_features
}
correlation_df = pd.DataFrame(list(correlations.items()), columns = ['Feature', 'Pearson Correlation'])
correlation_df.sort_values(by='Pearson Correlation', ascending=False)
```

```python
cat_features = [
    'is_female','is_smoker','region_northwest','region_southeast','region_southwest',
    'bmi_category_Normal','bmi_category_Overweight','bmi_category_Obese'
]
```

```python
from scipy.stats import chi2_contingency
import pandas as pd

alpha = 0.05

# Create bins for the target variable 'charges'
df_cleaned['charges_bin'] = pd.qcut(
    df_cleaned['charges'],
    q=4,
    labels=False
)

# Dictionary to store Chi-Square test results
chi2_results = {}

# Perform Chi-Square test for each categorical feature
for col in cat_features:
    contingency = pd.crosstab(
        df_cleaned[col],
        df_cleaned['charges_bin']
    )

    chi2_stat, p_val, _, _ = chi2_contingency(contingency)

    # Make decision based on significance level
    decision = (
        'Reject Null (Keep Feature)'
        if p_val < alpha
        else 'Accept Null (Drop Feature)'
    )

    chi2_results[col] = {
        'chi2_statistic': chi2_stat,
        'p_value': p_val,
        'Decision': decision
    }
    # Convert results dictionary into a DataFrame
chi2_df = pd.DataFrame(chi2_results).T

# Sort features by p-value
chi2_df = chi2_df.sort_values(by='p_value')

# Display the results
chi2_df
```

```python
x = final_df.drop('charges',axis = 1)
y = final_df['charges']
```

```python
x_train, x_test, y_train, y_test = train_test_split(x,y,test_size=0.20,random_state=42)
```

```python
from sklearn.linear_model import LinearRegression
```

**5. Modeling**

```python
model = LinearRegression()
model.fit(x_train,y_train)
```

```python
y_pred = model.predict(x_test)
```

**6. Model Evaluation**

```python
from sklearn.metrics import r2_score
r2 = r2_score(y_test,y_pred)
r2
n= x_test.shape[0]
p= x_test.shape[1]
adj_r2 = 1-(1-r2)*(n-1)/(n-p-1)
adj_r2
```
